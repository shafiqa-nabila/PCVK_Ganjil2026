```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
```

```python
img = cv.imread('ktm.jpg')
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
```

```python
h, w = img_rgb.shape[:2]
print(f"Ukuran asli: {w} x {h}")
```

```python
def rata_rata_intensitas(citra):
    return np.mean(citra)
```

```python
# 1. Hitung rata-rata intensitas citra asli
rata_asli = rata_rata_intensitas(img_rgb)
print(f"\nRata-rata intensitas citra asli: {rata_asli:.2f}")
```

```python
# 2. Kecilkan citra menjadi 1/4 dari ukuran asli
ukuran_kecil = (w//4, h//4)
img_kecil = cv.resize(img_rgb, ukuran_kecil)
rata_kecil = rata_rata_intensitas(img_kecil)
print(f"Rata-rata intensitas citra kecil: {rata_kecil:.2f}")
```

```python
# 3. Besarkan kembali ke ukuran semula
img_besar_kembali = cv.resize(img_kecil, (w, h))
rata_besar_kembali = rata_rata_intensitas(img_besar_kembali)
print(f"Rata-rata intensitas citra dibesarkan: {rata_besar_kembali:.2f}")
```

```python
# 4. Hitung selisih
selisih = abs(rata_besar_kembali - rata_asli)
print(f"\nSelisih rata-rata intensitas: {selisih:.2f}")
```

```python
# 5. Tampilkan visualisasi
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title(f'Citra Asli\n{rata_asli:.2f}')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img_kecil)
plt.title(f'Citra Kecil\n{rata_kecil:.2f}')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(img_besar_kembali)
plt.title(f'Citra Dibesarkan\n{rata_besar_kembali:.2f}')
plt.axis('off')

plt.suptitle('Perbandingan Rata-rata Intensitas', fontsize=14)
plt.tight_layout()
plt.show()
```

**Kesimpulan**

Mengecilkan lalu membesarkan kembali citra ke ukuran semula mengubah rata-rata intensitasnya karena proses interpolasi menyebabkan hilangnya informasi piksel asli yang tidak dapat dikembalikan.

```python
img1 = cv.imread('gunung.jpg')
img2 = cv.imread('pantai.jpg')
```

```python
img1_rgb = cv.cvtColor(img1, cv.COLOR_BGR2RGB)
img2_rgb = cv.cvtColor(img2, cv.COLOR_BGR2RGB)
```

```python
print("Ukuran citra 1 (gunung.jpg):", img1_rgb.shape)
print("Ukuran citra 2 (pantai.jpg):", img2_rgb.shape)
```

```python
def tambah_label(img, teks):
    h, w = img.shape[:2]

    # Tinggi bar adaptif (8% dari tinggi)
    tinggi_bar = max(20, int(h * 0.08))

    # Ukuran font adaptif
    skala_font = max(0.5, min(w, h) / 500)
    tebal = max(1, int(skala_font * 1.5))

    # Buat salinan
    hasil = img.copy()

    # Gambar bar hitam di bawah
    cv.rectangle(hasil, (0, h - tinggi_bar), (w, h), (0, 0, 0), -1)

    # Tambahkan teks di tengah
    ukuran_teks = cv.getTextSize(teks, cv.FONT_HERSHEY_SIMPLEX, skala_font, tebal)[0]
    x = (w - ukuran_teks[0]) // 2
    y = h - tinggi_bar + (tinggi_bar + ukuran_teks[1]) // 2

    cv.putText(hasil, teks, (x, y), cv.FONT_HERSHEY_SIMPLEX,
               skala_font, (255, 255, 255), tebal, cv.LINE_AA)

    return hasil
```

```python
img1_label = tambah_label(img1_rgb, "DOKUMENTASI PCVK - 2026")
img2_label = tambah_label(img2_rgb, "DOKUMENTASI PCVK - 2026")
```

```python
plt.figure(figsize=(14, 7))

plt.subplot(1, 2, 1)
plt.imshow(img1_label)
plt.title(f'Citra 1: gunung.jpg\n{img1_rgb.shape[1]}x{img1_rgb.shape[0]}')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img2_label)
plt.title(f'Citra 2: pantai.jpg\n{img2_rgb.shape[1]}x{img2_rgb.shape[0]}')
plt.axis('off')

plt.suptitle('Label Identitas Adaptif', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
```

```python
img1 = cv.imread('gunung.jpg')
img2 = cv.imread('pantai.jpg')
```

```python
img1_rgb = cv.cvtColor(img1, cv.COLOR_BGR2RGB)
img2_rgb = cv.cvtColor(img2, cv.COLOR_BGR2RGB)
```

```python
print("Ukuran citra 1 (gunung.jpg):", img1_rgb.shape)
print("Ukuran citra 2 (pantai.jpg):", img2_rgb.shape)
```

```python
def buat_thumbnail(img, ukuran=500, warna_bg=(255, 255, 255), nama_toko="TOKO KITA"):
    h, w = img.shape[:2]

    # Hitung rasio asli
    rasio = w / h

    # Tentukan ukuran setelah resize (pertahankan rasio)
    if rasio > 1:  # Lanskap (lebar > tinggi)
        new_w = ukuran
        new_h = int(ukuran / rasio)
    else:  # Potrait atau persegi (tinggi >= lebar)
        new_h = ukuran
        new_w = int(ukuran * rasio)

    # Resize gambar
    img_resize = cv.resize(img, (new_w, new_h))

    # Buat kanvas persegi dengan warna latar
    kanvas = np.full((ukuran, ukuran, 3), warna_bg, dtype=np.uint8)

    # Posisikan gambar di tengah
    y_offset = (ukuran - new_h) // 2
    x_offset = (ukuran - new_w) // 2
    kanvas[y_offset:y_offset+new_h, x_offset:x_offset+new_w] = img_resize
```

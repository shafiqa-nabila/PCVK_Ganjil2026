```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
imageurls = [
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/fruits.jpg",
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/butterfly.jpg",
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/starry_night.jpg",
]

for url in imageurls:
    image = io.imread(url) #read image
    image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image to half size
    image_2 =cv.cvtColor(image, cv.COLOR_BGR2RGB)  #convert color to RGB
    final_frame = cv.hconcat((image, image_2))  #concatenate image
    cv2_imshow(final_frame)    #show image
print('\n')
```

```python
image_vertikal = image_2.copy()

# Mencari titik tengah kolom
col = int(image_vertikal.shape[1] / 2)

# Jalan menyusuri baris untuk garis vertikal
for y in range(image_vertikal.shape[0]):
    image_vertikal[y, col] = [0, 255, 0]

final_frame = cv.hconcat((image_2, image_vertikal))
cv2_imshow(final_frame)
```

```python
image_diagonal = image_2.copy()

h, w = image_diagonal.shape[0], image_diagonal.shape[1]

# Garis diagonal utama (kiri atas ke kanan bawah) & silang (kanan atas ke kiri bawah)
for i in range(min(h, w)):
    image_diagonal[i, i] = [0, 255, 0] # Diagonal utama
    image_diagonal[i, w - 1 - i] = [0, 255, 0] # Diagonal menyilang

final_frame = cv.hconcat((image_2, image_diagonal))
cv2_imshow(final_frame)
```

```python
image_merah = image_2.copy()

# Mencari titik tengah baris (tinggi) dan kolom (lebar)
row = int(image_merah.shape[0] / 2)
col_center = int(image_merah.shape[1] / 2)

# Menentukan batas awal dan akhir garis (misal panjang garis 100 piksel)
start_x = col_center - 50
end_x = col_center + 50

# Menggambar garis horizontal berwarna merah [Blue=0, Green=0, Red=255]
for x in range(start_x, end_x):
    image_merah[row, x] = [0, 0, 255]

final_frame = cv.hconcat((image_2, image_merah))
cv2_imshow(final_frame)
```
